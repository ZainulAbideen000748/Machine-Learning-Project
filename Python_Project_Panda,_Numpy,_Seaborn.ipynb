**## Seaborn Library - Data Visualization - **

```python
# Aggregating sales and profit by category
cat_data = df.groupby('category')[['sales', 'profit']].sum().reset_index()

# Melting the dataset into a long-form format for seaborn
cat_plot = pd.melt(cat_data, id_vars=['category'], value_vars=['sales', 'profit'], var_name='metric', value_name='amount')

# Creating the bar plot
plt.figure(figsize=(8, 5))
sns.barplot(data=cat_plot, x="category", y="amount", hue="metric")
plt.title("Sales vs Profit by Category")
plt.xlabel("")
plt.ylabel("Amount ($)")
plt.show()
```

```python
# Create the 'order_year' column from 'order_date'
df['order_year'] = df['order_date'].dt.year

# Group by year and region to sum sales
yr_reg = df.groupby(["order_year", "region"], as_index=False)["sales"].sum()

plt.figure(figsize=(9, 5))
sns.lineplot(data=yr_reg, x="order_year", y="sales", hue="region", marker="o")
plt.xticks([2014, 2015, 2016, 2017])
plt.title("Yearly Sales by Region")
plt.ylabel("Sales ($)")
plt.xlabel("Year")
plt.show()
```

```python
# Grouping by sub-category to get total profit
sub = df.groupby('sub-category')[['profit']].sum().reset_index()
sub.columns = ['sub_category', 'total_profit']

sub_plot = sub.sort_values("total_profit").reset_index()
colors = dict(zip(sub_plot["sub_category"],
                  np.where(sub_plot["total_profit"] < 0, "crimson", "steelblue")))

plt.figure(figsize=(9, 6))
sns.barplot(data=sub_plot, y="sub_category", x="total_profit",
            hue="sub_category", palette=colors, legend=False)
plt.axvline(0, color="black", linewidth=1)
plt.title("Profit by Sub-Category (Red = Loss)")
plt.xlabel("Total Profit ($)")
plt.ylabel("")
plt.show()
```

```python
heat = df.pivot_table(values="profit", index="region", columns="category", aggfunc="sum")

plt.figure(figsize=(8, 4.5))
sns.heatmap(heat, annot=True, fmt=",.0f", cmap="RdYlGn", center=0)
plt.title("Profit Heatmap: Region x Category")
plt.show()
```
