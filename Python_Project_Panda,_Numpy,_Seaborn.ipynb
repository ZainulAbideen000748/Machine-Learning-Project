```python
df = df.drop(columns=["row_id", "country"], errors='ignore')
print(df)
```

```python
Profit_margin=df['profit']/df['sales']*100
print(Profit_margin)
```

```python
#Company KPIs / Financial Highlights
Total_sales=df['sales'].sum()
Total_Profit=df['profit'].sum()
margin=Total_Profit/Total_sales*100
Total_Orders=df['order_id'].count()
unique_orders=df['order_id'].nunique()
unique_customers=df['customer_name'].nunique()
return_rate=(df['returned']=='Yes').mean()*100
print(Total_sales)
print(Total_Profit)
print(margin)
print(Total_Orders)
print(unique_orders)
print(unique_customers)
print(return_rate)
```

```python
#Numpy Mathematical Calculations
sales_arr=df['sales'].to_numpy()
print(type(sales_arr))
print(sales_arr.shape)
```

```python
print('Mean :',np.mean(sales_arr).round(2))
print('Median :',np.median(sales_arr).round(2))
print('Standard Deviation :',np.std(sales_arr).round(2))
```
